```python
import pandas as pd
import matplotlib.pyplot as plt

# Đọc dữ liệu từ file CSV
data = pd.read_csv('data_processed.csv')

# Lấy top 10 quốc gia có số lần xuất hiện nhiều nhất
top_10_countries = data['Country'].value_counts().head(5)
print(top_10_countries)

# Tạo DataFrame chỉ chứa top 10 quốc gia này
top_10_countries_data = data[data['Country'].isin(top_10_countries.index)]

# Lọc các mục đã được đánh giá và chưa được đánh giá
rated_data = top_10_countries_data[top_10_countries_data['Rating'] != 'unknown']
unknown_data = top_10_countries_data[top_10_countries_data['Rating'] == 'unknown']

# Đếm số lượng mục đã được đánh giá và chưa được đánh giá trong mỗi quốc gia
rated_counts = rated_data['Country'].value_counts()
unknown_counts = unknown_data['Country'].value_counts()

# Tạo DataFrame từ các kết quả đếm
rating_summary = pd.DataFrame({
    'Rated': rated_counts,
    'Unknown': unknown_counts
}).fillna(0)

# Vẽ biểu đồ cột xếp chồng
plt.figure(figsize=(14, 8))
rating_summary.plot(kind='bar', color=['skyblue', 'lightcoral'], ax=plt.gca())

# Xoay các tên của cột x hướng chéo 60 độ 
plt.xticks(rotation=30)

# Thêm tiêu đề và nhãn
plt.title('Top 5 countries with the most occurrences ', fontsize=16)
plt.xlabel('Country', fontsize=14)
plt.ylabel('Count of Ratings', fontsize=14)
plt.legend(title='Rating Type', fontsize=12)

# Hiển thị biểu đồ
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Đọc dữ liệu từ file CSV
data = pd.read_csv('data_processed.csv')

# Lọc các quốc gia có số lượng count >= 2
filtered_countries = data['Country'].value_counts()
filtered_countries = filtered_countries[filtered_countries >= 2]

# Tạo DataFrame chỉ chứa các quốc gia đã lọc
filtered_countries_data = data[data['Country'].isin(filtered_countries.index)]

# Lọc các mục đã được đánh giá và chưa được đánh giá
rated_data = filtered_countries_data[filtered_countries_data['Rating'] != 'unknown']
unknown_data = filtered_countries_data[filtered_countries_data['Rating'] == 'unknown']

# Đếm số lượng mục đã được đánh giá và chưa được đánh giá trong mỗi quốc gia
rated_counts = rated_data['Country'].value_counts()
unknown_counts = unknown_data['Country'].value_counts()

# Tạo DataFrame từ các kết quả đếm
rating_summary = pd.DataFrame({
    'Rated': rated_counts,
    'Unknown': unknown_counts
}).fillna(0)

# Chỉ lấy 5 quốc gia mà số lượng "unknown" nhiều hơn số lượng "Rated"
top_5_countries_unknown_more = rating_summary[rating_summary['Unknown'] > rating_summary['Rated']].nlargest(5, 'Unknown')

# Vẽ biểu đồ cột xếp chồng
plt.figure(figsize=(14, 8))
top_5_countries_unknown_more.plot(kind='bar', color=['skyblue', 'lightcoral'], ax=plt.gca())

# Xoay các tên của cột x hướng chéo 60 độ 
plt.xticks(rotation=30)

# Thêm tiêu đề và nhãn
plt.title('Top 5 countries with the fewest occurrences ', fontsize=16)
plt.xlabel('Country', fontsize=14)
plt.ylabel('Count of Ratings', fontsize=14)
plt.legend(title='Rating Type', fontsize=12)

# Hiển thị biểu đồ
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Đọc dữ liệu từ file CSV
data = pd.read_csv('data_processed.csv')

# Thêm một cột mới để phân loại khoảng giá
bins = [0, 300000, 600000, 900000, 1200000, float('inf')]
labels = ['0-300k', '300k-600k', '600k-900k', '900k-1.2m', '1.2m+']
data['Price_Category'] = pd.cut(data['Price'], bins=bins, labels=labels, right=False)

# Lấy top 5 khoảng giá có số lần xuất hiện nhiều nhất
top_5_price_categories = data['Price_Category'].value_counts().head(5)
print(top_5_price_categories)

# Tạo DataFrame chỉ chứa top 5 khoảng giá này
top_5_price_categories_data = data[data['Price_Category'].isin(top_5_price_categories.index)]

# Lọc các mục đã được đánh giá và chưa được đánh giá
rated_data = top_5_price_categories_data[top_5_price_categories_data['Rating'] != 'unknown']
unknown_data = top_5_price_categories_data[top_5_price_categories_data['Rating'] == 'unknown']

# Đếm số lượng mục đã được đánh giá và chưa được đánh giá trong mỗi khoảng giá
rated_counts = rated_data['Price_Category'].value_counts()
unknown_counts = unknown_data['Price_Category'].value_counts()

# Tạo DataFrame từ các kết quả đếm
rating_summary = pd.DataFrame({
    'Rated': rated_counts,
    'Unknown': unknown_counts
}).fillna(0)

# Vẽ biểu đồ cột xếp chồng
plt.figure(figsize=(14, 8))
rating_summary.plot(kind='bar', color=['skyblue', 'lightcoral'], ax=plt.gca())

# Xoay các tên của cột x hướng chéo 60 độ 
plt.xticks(rotation=30)

# Thêm tiêu đề và nhãn
plt.title('Top 5 price ranges with the most occurrences', fontsize=16)
plt.xlabel('Price Range', fontsize=14)
plt.ylabel('Count of Ratings', fontsize=14)
plt.legend(title='Rating Type', fontsize=12)

# Hiển thị biểu đồ
plt.tight_layout()
plt.show()
```
